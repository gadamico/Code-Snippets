```python
def von_neumann_number(n: int) -> str:
    if n == 0:
        return '{}'
    else:
        return '{' + ','.join(von_neumann_number(j) for j in range(n)) + '}'
    
def zermelo_number(n: int) -> str:
    if n == 0:
        return '{}'
    else:
        return '{' + zermelo_number(n-1) + '}'
```

```python
from scipy import stats
```

Suppose we have an urn with 100 red balls, 100 green balls, and 100 black balls. What are the chances that we have no black balls in a draw of five balls?

We could work this out analytically. One way of thinking about is:

On the first pull we have 200 safe (i.e. non-black) balls out of 300. Then we'll have 199 safe balls out of 299. Etc.

$\large(\frac{200}{300}\large)\times\large(\frac{199}{299}\large)\times\large(\frac{198}{298}\large)\times\large(\frac{197}{297}\large)\times\large(\frac{196}{296}\large)$

```python
(200/300) * (199/299) * (198/298) * (197/297) * (196/296)
```

Another way of thinking about the problem is:

We could draw five from the 100 red balls and zero from the 100 green balls, or four from the red balls and one from the green, or three red and two green, or two red and three green, or one red and four green, or zero red and five green, and we have to manage one of these out of all the possible ways to draw five balls from a total of 300.

$\frac{\binom{100}{5}\binom{100}{0} + \binom{100}{4}\binom{100}{1} + \binom{100}{3}\binom{100}{2} + \binom{100}{2}\binom{100}{3} + \binom{100}{1}\binom{100}{4} + \binom{100}{0}\binom{100}{5}}{\binom{300}{5}}$

```python
# (100c5*100c0 + 100c4*100c1 + 100c3*100c2 + 100c2*100c3 + 100c1*100c4 + 100c0*100c5) / 300c5
((5 * 33 * 49 * 97 * 96)*2 + (25 * 33 * 49 * 97)*200 + (50 * 33 * 98)*100*99) / (5 * 299 * 149 * 297 * 296)
```

```python
# The values 0, 1, 2 will represent red, green, black
dist = stats.randint(0, 3)
```

```python
def check_for_black(n=100):
    dist = stats.randint(0, 3)
    ctr = 0
    for j in range(n):
        sample = dist.rvs(5)
        if 2 not in sample:
            ctr += 1
    return ctr / n
```

```python
results = []
for j in range(1000):
    results.append(check_for_black())
```

```python
np.mean(results)
```

```python
np.median(results)
```

```python
def gcf(m, n):
    
    """
    This function returns the greatest
    common factor of m and n.
    """
    
    m_fact = p_fact(m)
    n_fact = p_fact(n)
    
    short = min(len(m_fact), len(n_fact))
    long = max(len(m_fact), len(n_fact))
    
    out = []
    
    for j in range(short):
        out.append(min(m_fact[j], n_fact[j]))
        
    if sum(out) == 0:
        return 1
    
    return out
```

```python
gcf(2, 3)
```

```python
def gcf_alt(m, n):
    
    out = 1
    
    for j in range(2, min(m, n)//2 + 1):
        while m % j == 0 and n % j == 0:
            out *= j
            m = m/j
            n = n/j
    
    return out
```

```python
def lcm(m, n):
    
    out = 1
    
    for j in range(2, max(m, n) + 1):
        while m % j == 0 or n % j == 0:
            out *= j
            if m % j == 0:
                m = m/j
            if n % j == 0:
                n = n/j
    
    return out
```
